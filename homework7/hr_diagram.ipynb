```python

myarray = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows = 1)
print(myarray.shape)
#2642 rows
magnitude = myarray[:, 0]
temperature = myarray[:, 1]
#print(magnitude)
#print(temperature)
print(min(magnitude))
print(temperature[np.argmin(magnitude)])
#Omicron Persi
print(max(magnitude))
print(temperature[np.argmax(magnitude)])
#HIP 3829
print(min(temperature))
print(magnitude[np.argmin(temperature)])
#11 Leonis Minoris
print(max(temperature))
print(magnitude[np.argmax(temperature)])
#52 Sagitarii

log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
plt.figure(figsize = (8,6))
y = log_magnitude
x = log_temperature
plt.scatter(x, y, label = "Stars")

plt.plot(np.log(min(temperature)), np.log(magnitude[np.argmin(temperature)]), c = "orange", marker = "*", label = "Coldest")
plt.plot(np.log(max(temperature)), np.log(magnitude[np.argmax(temperature)]), c = "green", marker = "*", label = "Hottest")
plt.plot(np.log(temperature[np.argmin(magnitude)]), np.log(min(magnitude)), c = "purple", marker = "*", label = "Brightest")
plt.plot(np.log(temperature[np.argmax(magnitude)]), np.log(max(magnitude)), c = "black", marker = "*", label = "Dimmest")

plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.legend(loc = "lower left")
plt.show()
```
