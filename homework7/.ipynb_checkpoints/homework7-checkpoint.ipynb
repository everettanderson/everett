```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2*np.pi, 1000)

def makesinewave(x, A, w):
    y = A * np.sin(x * 2 * w)
    plt.plot(x, y)
print(makesinewave(x, 3, 1))
plt.show()
amplitude = [.5, 1, 1.5, 2, 2.5]
frequency = [1,2,3,4,5]



sinzip = zip(amplitude, frequency)
print(sinzip)
for amp, freq in sinzip:
    y = amp * np.sin(x * 2 * 1)
    plt.subplot(3,1,1)
    plt.title("Varying Amplitude")
    plt.xlabel("x")
    plt.ylabel("Asinx")
    plt.plot(x,y)
    y = 1 * np.sin(x*2*freq)
    plt.subplot(3,1,2)
    plt.title("Varying Frequency")
    plt.xlabel("x")
    plt.ylabel("sinBx")
    plt.plot(x,y)
    y = amp * np.sin(x * 2 * freq)
    plt.subplot(3, 1, 3)
    plt.title("Varying both simultaneously")
    plt.xlabel("x")
    plt.ylabel("CsinDx")
    plt.plot(x, y)
plt.suptitle("Varying Amplitude/Frequency/Both")
plt.show()
```
